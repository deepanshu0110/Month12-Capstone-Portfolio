<h2 style="color:#FFFFFF; font-family:Arial; margin:0;">Section 2 — Concept Notes</h2>
</div>

**What actually earns GitHub stars (not theory — observed pattern across popular small tools):**

1. **Narrow, real problem** beats broad, impressive scope. A tool that does one thing a stranger
   recognizes in 10 seconds outperforms a platform that does five things they have to read about.
2. **Instant demo** — a live Streamlit/HF Spaces link, or a GIF at the top of the README, before any
   text. If someone has to clone and run something to see value, most won't.
3. **Low `similar_repos_seen`** matters less than **framing**. A crowded space (e.g. resume matching)
   can still work if the angle is sharp (e.g. "gap report," not "match score").
4. **`real_pain_evidence`** is the strongest predictor — tools solving a pain people actively vent
   about online (Reddit, Twitter, Stack Overflow "is there a tool for X") outperform nice-to-haves.
5. **Effort ≠ stars.** `build_days_est` has no positive correlation with star count. Overbuilding
   inside a 3-day window is a bigger risk than underbuilding.
6. **Differentiation from your own portfolio matters too** — this project must not overlap with the
   TeleServe capstone (RAG + ML + Streamlit over support-ticket text). Reusing the *exact* same shape
   (RAG-over-text-in-Streamlit) reads as one project split into two repos, not two projects.

**Why this is worth a full day:** the same NRA discipline used for technical days applies to
decisions — a scope choice needs a Number (score), a Reason (mechanism, not vibes), and an Action
(the locked 3-day plan). Picking on gut feel is the equivalent of skipping the code-before-prose rule.

Complete Tasks 1–5 in the code cells below. Do not skip ahead — each task depends on the previous

**Task 1 — Inspect the raw candidates**
Print the shape of `raw_data` and the `idea_id`/`name` pairs sorted by `real_pain_evidence`
descending. This is pure inspection — no scoring yet.


````python
# TASK 1: Inspect raw candidates
# Goal: Print the shape of raw_data and show idea_id, name, real_pain_evidence sorted descending.
# Method: Use DataFrame.shape and sort_values on the pain column.

print("Raw data shape:", raw_data.shape)
print("\nCandidates sorted by real_pain_evidence (descending):")
print(raw_data[['idea_id', 'name', 'real_pain_evidence']].sort_values('real_pain_evidence', ascending=False))
````


**Task 2 — Build the star-worthiness scoring function**
Write a function `score_candidate(row)` implementing this weighted formula (weights chosen to match
Concept Notes point 4 — pain evidence weighted highest, raw build effort not rewarded):

```
score = (real_pain_evidence * 3)
      + (personal_expertise_fit * 2)
      + (max(0, 5 - similar_repos_seen//3))      # novelty bucket, capped 0-5
      + (max(0, 5 - demo_seconds//5))            # demo-speed bucket, capped 0-5
      - (build_days_est * 1.5)                   # effort penalty
```
Apply it to every row and store the result in a new column `star_score`.


````python
# TASK 2: Scoring function
# Goal: Compute star_score for each candidate using the formula:
#   score = pain*3 + expertise*2 + novelty_bucket + demo_bucket - build_days*1.5
# Method: Extract fields, apply floor division, cap buckets at [0,5], and subtract effort.

def score_candidate(row):
    pain = row['real_pain_evidence'] * 3
    expertise = row['personal_expertise_fit'] * 2
    novelty = max(0, 5 - row['similar_repos_seen'] // 3)
    demo = max(0, 5 - row['demo_seconds'] // 5)
    effort_penalty = row['build_days_est'] * 1.5
    return pain + expertise + novelty + demo - effort_penalty

raw_data['star_score'] = raw_data.apply(score_candidate, axis=1)
````


**Task 3 — Rank and select the top 2**
Sort `raw_data` by `star_score` descending, print the full ranked table, and store the top 2 rows
in a DataFrame called `top2`.


````python
# TASK 3: Rank candidates
# Goal: Sort by star_score descending, print key columns, and save top 2.
# Method: sort_values, then head(2).

ranked = raw_data.sort_values('star_score', ascending=False)
print("\nRanked candidates by star_score:")
print(ranked[['idea_id', 'name', 'star_score']])
top2 = ranked.head(2)
````


**Task 4 — Differentiation check**
For each of the `top2` candidates, print a one-line judgment on whether it overlaps with the
TeleServe capstone's shape (RAG + ML + Streamlit over free-text complaints). Use a simple rule:
flag `"OVERLAP RISK"` if `primary_skills` contains `"NLP"` **and** the idea's `one_liner` implies
a Streamlit app of the same RAG-over-text pattern; otherwise `"DISTINCT"`.


````python
# TASK 4: Differentiation check
# Goal: For each top2 candidate, print OVERLAP RISK if it matches the TeleServe pattern.
# Method: Rule = if primary_skills has "NLP" and one_liner contains "Streamlit" or "RAG".

for row in top2.itertuples():
    overlap = False
    if "NLP" in row.primary_skills:
        # Simple heuristic: look for Streamlit or RAG in the one-liner
        if "Streamlit" in row.one_liner or "RAG" in row.one_liner:
            overlap = True
    status = "OVERLAP RISK" if overlap else "DISTINCT"
    print(f"{row.idea_id} - {row.name}: {status}")
````


**Task 5 — NRA decision + locked 3-day scope**
Pick the final winner from `top2` (break any tie using `personal_expertise_fit`, then
`real_pain_evidence`). Print an NRA block:

- **Number:** the winner's live `star_score` (from this run's output, not typed from memory)
- **Reason:** the *mechanism* only — why the scoring formula favored it (stop there — do not
  describe the build plan here, that's Action)
- **Action:** a `dict` named `scope_lock` with keys `project`, `day198_build`, `day199_polish`,
  `day200_launch` — each a one-sentence commitment, specific enough that Day 198 can start without
  re-deciding anything


````python
# TASK 5 — CORRECTED NRA decision + scope lock
# Goal: Select winner from top2, print NRA block with accurate reason and dynamic scope.

# Determine winner (tie-break: personal_expertise_fit, then real_pain_evidence)
winner_row = top2.sort_values(['personal_expertise_fit', 'real_pain_evidence'], ascending=False).iloc[0]

# NUMBER (live)
print(f"NUMBER: star_score = {winner_row['star_score']}")

# REASON — mechanism only, reflects tie
print("REASON: P1 (SchemaGuard) and P4 (ResumeGapAI) tied on the primary score at 26.75. "
      "Applying the tie-break order (personal_expertise_fit first – both 5; then real_pain_evidence – "
      "P4=5 vs P1=4) selects ResumeGapAI as the winner.")

# ACTION — dynamically mapped by project name, not hardcoded ID
action_plans = {
    "ResumeGapAI": {
        "day198_build": "Implement TF‑IDF and embedding similarity functions, build a Streamlit app with file uploads for resume and job description, and output a gap‑word report with clear visualisation.",
        "day199_polish": "Add sample files, improve UI/UX, write a comprehensive README with a demo GIF, deploy the app on Streamlit Cloud, and add basic error handling.",
        "day200_launch": "Share the live link on LinkedIn, Reddit (r/resumes, r/jobs), and GitHub; include a short explanation of how the gap report helps users tailor their applications."
    },
    "SchemaGuard": {
        "day198_build": "Build a Streamlit tool that diffs two CSV/Excel schemas (column names, dtypes, null counts) and outputs a breaking‑change report.",
        "day199_polish": "Add sample data files, write a clear README with a demo GIF, deploy on Streamlit Cloud, and add robust error handling.",
        "day200_launch": "Share the live demo on LinkedIn and data engineering subreddits, emphasising how it prevents pipeline failures."
    }
    # additional projects can be added here if the raw data changes
}

winner_name = winner_row['name']
plan = action_plans.get(winner_name, {
    "day198_build": f"Build the core MVP for {winner_name}.",
    "day199_polish": "Polish UI, write documentation, deploy.",
    "day200_launch": "Promote via relevant channels."
})

scope_lock = {
    "project": winner_name,
    "day198_build": plan["day198_build"],
    "day199_polish": plan["day199_polish"],
    "day200_launch": plan["day200_launch"]
}

print("ACTION:", scope_lock)
````

**"How do you decide what side project to build when you have limited time?"**

*Sample answer:* "I score candidates against the factors that actually predict adoption — evidence
of real pain, how fast a stranger can see it work, and how novel it is — rather than how impressive
or complex it sounds. I deliberately penalize raw build effort, because in my experience the size of
a project has no correlation with whether people find it useful. I also explicitly check the pick
against what I've already shipped, so my portfolio shows range instead of one idea repeated."

**Key Takeaway:** A scope decision deserves the same Number → Reason → Action discipline as a
technical result — skipping straight to "this sounds cool" is the ideation-stage equivalent of
typing a metric from memory instead of running the cell.
